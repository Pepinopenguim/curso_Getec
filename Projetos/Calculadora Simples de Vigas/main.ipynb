```python
import indeterminatebeam as ib

viga = ib.Beam(10)

viga.add_supports(
    ib.Support(0, (1, 1, 0)),
    ib.Support(10, (1, 0, 1)),
    ib.Support(7, (0, 1, 0)),
)
```

```python
# Determinar esforcos
forca_pontual = ib.PointLoadV(-1000, 5)
forca_distribuida = ib.DistributedLoadV(-500, (0, 4))
torque_pontual = ib.PointTorque(2000, 8)

viga.add_loads(forca_pontual, forca_distribuida, torque_pontual)
```

```python
viga.analyse()
```

```python
figura1 = viga.plot_beam_external()
figura1
```

```python
cortantes = viga.plot_shear_force()
cortantes
```

```python
momentos = viga.plot_bending_moment()
momentos
```
